# Exno: 1 – Employee Data Analysis using Pandas and NumPy

This notebook reproduces the employee-data operations in the supplied Colab notebook.

## 1. Import libraries and create the employee dataset

```python
import pandas as pd
import numpy as np
```

```python
emp = {
    "Emp_ID": ["E101", "E102", "E103", "E104", "E105", "E106", "E107", "E108", "E109", "E110"],
    "Name": ["Arjun", "Priya", "Karthik", "Divya", "Rahul", "Sneha", "Vignesh", "Ananya", "Sanjay", "Meena"],
    "Dept": ["IT", "HR", "Finance", "IT", "Marketing", "HR", "Finance", "IT", "Sales", "Marketing"],
    "Age": [25, 28, np.nan, 24, 32, 27, 35, np.nan, 26, 31],
    "Salary": [35000, 42000, 50000, np.nan, 55000, 40000, 65000, 48000, np.nan, 52000],
    "Experience": [2, 4, 6, 1, np.nan, 3, 10, 5, 2, 6],
    "City": ["Chennai", "Bangalore", "Coimbatore", "Chennai", "Mumbai", np.nan, "Salem", "Bangalore", "Chennai", "Trichy"]
}
df = pd.DataFrame(emp)
print(df)
```

## 3. Check and handle missing values

```python
df.isnull()
```

```python
df.isnull().sum()
```

```python
df.isnull().any()
```

```python
df["Salary"] = df["Salary"].fillna(df["Salary"].mean())
```

```python
df["Experience"] = df["Experience"].fillna(0)
```

```python
df
```

```python
df.isnull().sum()
```

## 4. Rename a column and calculate salary values

```python
df.rename(columns={"Salary": "Monthly_Salary"})
```

```python
df["Annual_Salary"] = df["Salary"] * 12
```

```python
df["Bonus"] = df["Salary"] * 0.10
```

```python
df
```

## 5. Filter and group employee records

```python
df[df["Salary"] > 5000]
```

```python
df[(df["Salary"] > 5000) & (df["Dept"] == "IT")]
```

```python
df.groupby("Dept")["Salary"].mean()
```

```python
df["Dept"].unique()
```

```python
df.groupby("Dept")["Salary"].max()
```

```python
df["Dept"].value_counts()
```

## 6. NumPy operations and derived columns

```python
np.isnan(df["Age"])
```

```python
np.mean(df["Salary"])
```

```python
np.nanmax(df["Salary"])
```

```python
df["Bonus"] = np.round(df["Salary"] * 0.10, 3)
df
```

```python
df["Performance"] = np.where(df["Salary"] > 50000, "Excellent", "Good")
df
```

```python
df["Grade"] = np.select(
    [df["Salary"] >= 60000, df["Salary"] >= 50000, df["Salary"] >= 40000],
    ["A", "B", "C"],
    default="D"
)
df
```

```python
df["New Salary"] = np.multiply(df["Salary"], 1.10)
df
```

```python
df["Experience Months"] = np.multiply(df["Experience"], 12)
df
```
